## 3. Polynomial Regression

### 공부 시간에 따른 시험 점수 ( 우등생)

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('PolynomialRegressionData.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:,-1].values
```

## 3-1 단순 선형 회귀(Simple Linear Regression)

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression() # 객체 생성 
reg.fit(X,y) # 학습을 통한 모델 생성
```

### 데이터 시각화 (전체)

```python
plt.scatter(X, y, color='blue') # 산점도 그래프 생성
plt.plot(X, reg.predict(X), color='green') # 선그래프 생성
plt.title('Score by hours (genius)') # 제목
plt.xlabel('hours') # X축 이름
plt.ylabel('Score') # Y축 이름
plt.show()
```

```python
reg.score(X, y) # 전체 데이터를 통한 모델 평가
```

## 3-2 다항 회귀 (Polynomial Regression)

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=4) # 2차
X_poly = poly_reg.fit_transform(X)
X_poly[:5] # [X] -> [X^0 , X^1, X^2] -> X가 3이라면 [1, 3, 9]로 변환
```

```python
X[:5]
```

```python
poly_reg.get_feature_names_out()
```

```python
lin_reg = LinearRegression() # 객체 생성 
lin_reg.fit(X_poly, y) # 변환된 X와 y를 가지고 모델 생성(학습)
```

### 데이터 시각화 (변환된 X와 y)

```python
plt.scatter(X, y, color='blue') # 산점도 그래프 생성
plt.plot(X, lin_reg.predict(poly_reg.fit_transform(X)), color='green') # 선그래프 생성
plt.title('Score by hours (genius)') # 제목
plt.xlabel('hours') # X축 이름
plt.ylabel('Score') # Y축 이름
plt.show()
```

```python
# 선을 조금 더 부드럽게 만드는 과정
X_range = np.arange(min(X), max(X), 0.1) # X의 최소값에서 최대값까지의 범위를 0.1 단위로 잘라서 데이터를 생성
X_range
```

```python
X_range.shape
```

```python
X_range = X_range.reshape(-1, 1) # row 개수는 자동으로 계산, Column 개수는 1개
X_range.shape
```

```python
X_range[:5]
```

```python
plt.scatter(X, y, color='blue') # 산점도 그래프 생성
plt.plot(X_range, lin_reg.predict(poly_reg.fit_transform(X_range)), color='green') # 선그래프 생성
plt.title('Score by hours (genius)') # 제목
plt.xlabel('hours') # X축 이름
plt.ylabel('Score') # Y축 이름
plt.show()
```
