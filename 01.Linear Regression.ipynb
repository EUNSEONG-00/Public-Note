# 1. Linear Regression

공부 시간에 따른 시험 점수

```python
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import plotly.express as px
import scipy.stats as stats

mpl.rc('font', family='Malgun Gothic')
```

```python
dataset= pd.read_csv('LinearRegressionData.csv')
```

```python
dataset.head()
```

```python
X = dataset.iloc[:,:-1].values # 처음부터 마지막 컬럼 직전까지의 데이터(독립변수 - 원인)
y = dataset.iloc[:, -1].values # 마지막 컬럼의 데이터 (종속변수 - 결과)
```

```python
X, y
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression() # 객체 생성 
reg.fit(X,y) # 학습을 통한 모델 생성
```

```python
y_pred = reg.predict(X) # x에 대한 예측 값 생성
y_pred
```

```python
plt.scatter(X, y, color='blue') # 산점도 그래프 생성
plt.plot(X, y_pred, color='green') # 선그래프 생성
plt.title('Score by hours') # 제목
plt.xlabel('hours') # X축 이름
plt.ylabel('Score') # Y축 이름
plt.show()
```

```python
print('9시간 공부했을 때 예상 점수: ', reg.predict([[9]])) # 2차원 배열 형태이기 때문에, 대괄호 2번 사용
```

```python
reg.coef_ # 기울기 (m)
```

```python
reg.intercept_ # y절편 (b)
```

### 데이터 세트 분리


```python
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import plotly.express as px
import scipy.stats as stats

mpl.rc('font', family='Malgun Gothic')
```

```python
dataset= pd.read_csv('LinearRegressionData.csv')
dataset
```

```python
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0) # 훈련 80 : 테스트 20 으로 분리
```

```python
X, len(X) # 전체 데이터 X 개수
```

```python
X_train, len(X_train) # 훈련 세트 X 개수
```

```python
X_test, len(X_test) # 테스트 세트 X 개수
```

```python
y, len(y) # 전체 데이터 y 개수
```

```python
y_train, len(y_train) # 훈련 세트 y 개수
```

```python
y_test, len(y_test) # 테스트 세트 y 개수
```

### 분리된 데이터를 통한 모델링

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
```

```python
reg.fit(X_train, y_train) # 훈련 세트로 학습
```

### 데이터 시각화(훈련 세트)

```python
plt.scatter(X_train, y_train, color='blue') # 산점도 그래프 생성
plt.plot(X_train, reg.predict(X_train), color='green') # 선그래프 생성
plt.title('Score by hours (train data)') # 제목
plt.xlabel('hours') # X축 이름
plt.ylabel('Score') # Y축 이름
plt.show()
```

### 데이터 시각화(테스트 세트)

```python
plt.scatter(X_test, y_test, color='blue') # 산점도 그래프 생성
plt.plot(X_train, reg.predict(X_train), color='green') # 선그래프 생성, 훈련 테스트 시각화와 동일하게 둠 
plt.title('Score by hours (test data)') # 제목
plt.xlabel('hours') # X축 이름
plt.ylabel('Score') # Y축 이름
plt.show()
```

```python
reg.coef_
```

```python
reg.intercept_
```

### 모델 평가

```python
reg.score(X_test, y_test) # 테스트 세트를 통한 모델 평가, 1에 가까울 수록 높은 평가 점수 
```

```python
reg.score(X_train, y_train) # 훈련 세트를 통한 모델 평가 
```

## 경사하강법(Gradient Descent)

max_iter : 훈련 세트 반복 횟수 (Epoch 횟수)

eta0 : 학습률 (learning rate)

```python
from sklearn.linear_model import SGDRegressor # SGD : Stochastic Gradient Descent 확률적 경사 하강법 

# 지수표기법
# 1e-3 : 0.001 (10^-3)
# 1e-4 : 0.0001 (10^-4)
# 1e+3 : 1000 (10^3)
# 1e+4 : 10000 (10^4)

# sr = SGDRegressor(max_iter=200, eta0=1e-4, random_state=0, verbose=1) # loss가 최소화 됨 
sr = SGDRegressor()
sr.fit(X_train, y_train)
```

```python
plt.scatter(X_train, y_train, color='blue') # 산점도 그래프 생성
plt.plot(X_train, sr.predict(X_train), color='green') # 선그래프 생성
plt.title('Score by hours (train data, SGD)') # 제목
plt.xlabel('hours') # X축 이름
plt.ylabel('Score') # Y축 이름
plt.show()
```

```python
sr.coef_, sr.intercept_
```

```python
sr.score(X_test, y_test) # 테스트 세트를 통한 모델 평가
```

```python
sr.score(X_train, y_train) # 훈련 세트를 통한 모델 평가
```
