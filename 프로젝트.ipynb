# 프로젝트 소개

###  지금부터 여러분은 대형 마트 기업에 막 입사한 신입 데이터 분석가입니다. 첫 업무는 2년 치 고객 데이터를 활용해 고객을 여러 그룹으로 분류하고 이에 따라 마케팅 전략을 검토해 보는 거예요. 고객 그룹을 나눈 뒤 매출에 크게 기여하는 핵심 고객 그룹을 찾아 그들의 특성과 소비 성향을 분석해 봅시다. 분석 결과를 바탕으로 핵심 고객 그룹을 집중 공략하면 한정된 마케팅 자원 안에서 최선의 결과를 얻을 수 있겠죠?

### 고객을 여러 그룹으로 나누는 방법에는 대표적으로 RFM 분석이 있습니다. 얼마나 최근에 상품을 구매했는지, 얼마나 자주 상품을 구매했는지, 그리고 얼마나 돈을 많이 썼는지에 따라 고객을 분류하는 방법이죠. 주어진 고객 데이터에서 이러한 정보를 모두 파악할 수 있으니 RFM 분석을 활용해 업무를 완수해 봅시다.

### 우선 고객 데이터를 탐색하고 전처리하는 과정이 필요해요. 이후에는 원하는 형태로 데이터를 가공하면서 필요한 정보를 추출하고 그래프도 그려 보며 인사이트를 얻어야 하죠. 기본적인 파이썬 문법과 데이터 분석에 관한 지식을 잘 익혀 뒀다면 어렵지 않을 겁니다.

## RFM 분석:


RFM은 Recency, Frequency, Monetary 각각의 앞 글자를 따서 만들어진 약자예요. 이 세 항목의 의미는 다음과 같습니다.  

Recency: 고객이 얼마나 최근에 상품을 구매했는가?  
Frequency: 고객이 얼마나 자주 상품을 구매했는가?  
Monetary: 고객이 상품 구매에 얼마나 많은 돈을 썼는가?  

 RFM 분석을 한다는 건 위 세 항목을 토대로 고객의 가치를 판단해 고객을 분류한다는 뜻입니다. 더 최근에 구매할수록, 더 자주 구매할수록, 그리고 돈을 더 많이 쓸수록 가치가 큰 고객으로 여기는 거죠.  

 한번 예를 들어 볼게요. 최근 1년 동안의 데이터를 기반으로 두 고객의 구매 패턴을 비교해 봅시다. 고객 A는 고객 B보다 더 최근에 상품을 구매했네요. 그리고 고객 A는 고객 B에 비해 1년 동안 구매 횟수도 많고 구매 금액도 더 큽니다. Recency, Frequency, Monetary 어느 요소로 보든 고객 A가 고객 B보다 더 가치 있는 고객이라고 판단할 만하죠.  

고객	마지막 구매일	구매 횟수	구매 금액
  A	       3일 전	     10회	   100만 원
  B	     100일 전	      1회	     5만 원
 RFM 분석은 아주 기본적이지만 효과적으로 고객을 분류할 수 있는 방법이라 알려져 있습니다. 이렇게 분류된 고객 그룹 하나하나를 세그먼트(Segment)라고 하는데요. PM이나 마케터는 고객 세그먼트에 따라 서로 다른 고객 관리 전략을 수립하고 적용해 볼 수 있습니다.

***RFM 분석을 통해 고객 분류하기***
Recency, Frequency, Monetary 항목에 등급 부여하기
 RFM 분석을 하려면 우선 Recency, Frequency, Monetary와 관련된 데이터가 있어야겠죠. 일반적으로 Recency는 기준 시점부터 가장 최근 구매일이 며칠 전인지를 나타냅니다. Frequency는 기준 기간 동안 상품을 구매한 횟수, Monetary는 기준 기간 동안 구매한 상품의 총금액을 의미하고요. 예를 들어 아래와 같은 데이터가 마련되어 있어야 RFM 분석이 가능합니다.

고객	Recency	Frequency	Monetary
  A	    3일	    10회	    100만 원
  B	  100일	     1회	      5만 원
  C	   40일	     3회	     20만 원
  D	    5일	     7회	     50만 원
  
 그 다음에는 Recency, Frequency, Monetary 각 항목에 등급을 부여해야 합니다. 등급을 매기는 데에는 크게 세 가지 방법이 있는데요. 각 등급마다 고객의 수가 비슷해지도록 하거나, 구간의 길이를 똑같이 나누거나, 아니면 임의로 구간을 설정할 수도 있습니다. 임의로 구간을 설정하는 예시를 하나 들어 볼게요. 각 항목의 등급을 3개로 나누되 아래처럼 기준을 정해 등급을 나누겠습니다. 등급은 숫자가 클수록 더 좋은 고객을 뜻합니다.

Recency: 7일 전까지는 3등급, 8일 전부터 30일 전까지는 2등급, 그보다 과거는 1등급
Frequency: 10회 이상은 3등급, 4회 이상 10회 미만은 2등급, 3회 이하는 1등급
Monetary: 100만 원 이상은 3등급, 20만 원 이상 100만 원 미만은 2등급, 20만 원 미만은 1등급
그러면 위의 데이터에는 이런 식으로 등급이 매겨질 거예요.

고객	Recency	Recency 등급	Frequency	Frequency 등급	Monetary	Monetary 등급
  A	    3일	        3	     10회	         3	        100만 원	        3
  B	  100일	        1	      1회	         1	          5만 원	        1
  C	   33일	        1	      3회	         1	        150만 원	        3
  D	    5일	        3	      7회	         2	         15만 원	        1
  
***등급의 조합으로 세그먼트 나누기***

이렇게 Recency, Frequency, Monetary 각 항목의 등급을 구하고 나면 이 셋을 조합하여 고객 세그먼트를 나눌 수 있습니다. 위 예시에서는 항목마다 등급이 3개씩이니 3x3x3=27개의 세그먼트가 나올 수 있죠.

고객 세그먼트	Recency 등급	Frequency 등급	Monetary 등급
       1	       3	         3	                3
       2	       3	         3	                2
       3	       3	         3	                1
      …	          …	          …	                …
      25	       1	         1	                3
      26	       1	         1               	2
      27	       1             1	                1 
      
 이제 세그먼트별로 다른 전략을 세워 고객을 관리할 수 있습니다. 모든 등급이 높은 고객은 절대 놓치면 안 되니 수시로 쿠폰을 발급하거나 이벤트를 진행해 꾸준히 높은 매출을 올리게 해야 합니다. 반면 등급이 모두 낮은 고객은 이미 이탈했지만 딱히 아쉬울 것도 없는 고객이죠. Monetary 등급만 높고 Recency, Frequency 등급이 낮은 고객은 조금 더 자주 구매를 하도록 특별 쿠폰과 같은 방안을 모색해 보면 좋을 거예요. 반대로 Recency, Frequency 등급은 높은데 Monetary 등급이 낮은 고객은 충성도가 검증된 셈이니 상품을 맞춤 추천하여 추가 구매를 유도해 볼 수 있겠습니다.

등급의 조합으로 고객 세그먼트를 나누면 세그먼트 개수는 이론적으로 등급 개수의 세제곱이 됩니다. 예시에서는 등급을 3개로 두어 27개의 세그먼트가 나왔는데요. 만약 등급을 5개로 둔다면 125개의 세그먼트가 나올 겁니다. 세그먼트 수가 많으면 고객 그룹을 더 세부적으로 파악하고 타깃을 정밀하게 설정할 수 있지만, 그만큼 세그먼트마다 고객을 관리하고 맞춤형 전략을 짜는 건 더 복잡해집니다. 일부 세그먼트는 너무 작은 규모여서 별로 의미가 없을 수도 있고, 비슷한 특성을 가진 고객이 서로 다른 세그먼트에 속하게 될지도 모르죠.

가중합으로 고객 지표를 구해 세그먼트 나누기
이러한 한계를 보완하고자 Recency, Frequency, Monetary 각 등급으로 가중합을 구해 고객 지표를 계산한 뒤, 이를 기반으로 다시 등급을 나누어 더 큰 단위의 고객 세그먼트를 구하기도 합니다. 먼저 고객 지표를 구하는 방법부터 살펴볼게요. 아래 식처럼 항목마다 가중치와 등급 값을 곱하여 더해 주면 됩니다.

**고객 지표 =wR∗R+wF∗F+wM∗M**


(wR: Recency 가중치, wF: Frequency 가중치, wM: Monetary 가중치, R: Recency 등급, F: Frequency 등급, M: Monetary 등급)

이때 가중치는 아래와 같은 조건을 따라야 합니다.

(wR>=0,wF>=0,wM>=0,wR+wF+wM=1) 

 가중치는 Recency, Frequency, Monetary 각 항목의 중요도를 고려하여 설정할 수 있습니다. 가령 배달 앱처럼 주문 금액의 상한이 그리 높지 않고 자주 이용하는 게 중요한 서비스에서는 Monetary에 비해 Recency나 Frequency가 더 중요한 항목이겠죠. 그러면 Monetary 가중치를 낮추고 다른 두 가중치를 높여서 고객 지표를 구하면 됩니다. 한편 가전제품은 주문당 결제 금액이 크고 한번 구매를 하면 재구매까지 시간이 오래 걸릴 테니, 판매사 입장에서는 다른 항목보다 Monetary 가중치를 상대적으로 높게 설정하는 게 좋을 거예요.

이렇게 계산한 고객 지표를 적절한 구간으로 나누어 최종 RFM 등급을 구할 수 있습니다. RFM 등급이 곧 고객 세그먼트가 되겠죠. 고객 세그먼트를 몇 개로 나눌지는 직접 결정하면 됩니다. 이를테면 고객 지표의 범위가 0점에서 5점 사이로 나온다고 가정할게요. 세그먼트를 4개로 나누기로 하고, 4점 이상이면 VIP 등급, 3점 이상 4점 미만이면 골드 등급, 2점 이상 3점 미만이면 실버 등급, 그 밑으로는 브론즈 등급으로 분류해 볼 수 있겠죠. 고객 세그먼트가 수십 개씩 나올 때에 비해 복잡성이 줄고 맞춤형 전략을 수립하기도 쉬워질 겁니다.

그럼 예시 데이터에서 고객 지표를 구하고 RFM 등급까지 부여해 세그먼트를 나눠 보겠습니다. 우선 Recency, Frequency, Monetary 세 항목이 똑같이 중요하다 가정할게요. 세 항목의 중요도가 전부 동일하다면 가중치 조건에 따라 세 가중치는 모두 1/3이 되어야 합니다. 이 가중치 세트로 고객 지표를 계산해 봅시다. 예를 들어 고객 A의 고객 지표는 1/3 * 3 + 1/3 * 3 + 1/3 * 3 = 3이 되는 거죠. 동일한 방식으로 나머지 고객들도 계산하면 다음과 같은 결과가 나옵니다.

고객	Recency 등급	Frequency 등급	Monetary 등급	RFM 고객 지표
A	3	3	3	3    
B	1	1	1	1  
C	1	1	3	5/3  
D	3	2	1	2  
이제 고객 지표에서 RFM 등급을 구할 차례입니다. RFM 등급 개수는 3개로 하고, 고객 지표가 1 이상 2 미만이면 1등급, 2 이상 3 미만이면 2등급, 3이면 3등급이라고 구간을 정해 볼게요. 그러면 RFM 등급이 아래처럼 매겨지게 됩니다. 예시 데이터에는 고객이 4명뿐이지만 실제로 수천 명 또는 수만 명의 고객이 단 3개의 세그먼트로 분류된다면 고객 관리가 훨씬 수월해지겠죠?  

고객	RFM 고객 지표	RFM 등급
A	3	3
B	1	1
C	5/3	1
D	2	2


**매출 기여도를 고려한 가중치 조정**

 앞서 가중치를 설정할 때 Recency, Frequency, Monetary 중에 더 비중을 두고 싶은 항목이 무엇인지에 따라 가중치를 설정하라고 했는데요. 이렇게 가중치 세트를 적용했는데 고객 세그먼트가 뚜렷하게 분류되지 않았다면 가중치를 조정하고 세그먼트를 재구성하는 게 좋습니다.  

여기서 뚜렷함의 기준이 되는 건 일반적으로 매출 기여도입니다. 매출 기여도는 말 그대로 매출에 얼마나 기여하는지를 나타내는 수치예요. 한 세그먼트의 매출 기여도는 전체 매출 대비 해당 세그먼트 고객의 매출 비율로 구할 수 있죠. 세그먼트마다 매출 기여도를 구해 보고 매출 기여도가 뚜렷하게 차이가 나는지, 아니면 모든 세그먼트의 매출 기여도가 비슷비슷한지 확인하면 됩니다.  

만약 VIP, 골드, 실버, 브론즈로 고객 등급을 분류했다면 VIP 등급의 매출 기여도가 확연히 높고 아래 등급으로 내려갈수록 매출 기여도가 낮아져야 좋겠죠. 그만큼 VIP 고객들이 특별한 가치가 있다는 뜻이고 이들을 집중적으로 관리하면 될 테니까요. 한편 각 등급의 매출 기여도가 비슷한 수준이라면 제대로 분류가 되지 않았다는 뜻이며 어느 세그먼트에 고객 관리 역량을 집중해야 할지 결정하기가 어려울 겁니다.  

가전제품 판매 회사에서 1년마다 RFM 분석을 통해 고객을 VIP, 골드, 실버, 브론즈 네 등급으로 분류하여 관리한다고 가정해 봅시다. 지난 1년의 데이터에 Recency, Frequency, Monetary 각 가중치를 1/3씩 균등하게 적용하여 세그먼트를 나눴더니 아래와 같은 결과가 나왔다고 해요. VIP 고객의 매출 기여도가 가장 높긴 하지만 브론즈 고객과의 차이가 6% 포인트에 불과합니다. 세그먼트마다 매출 기여도 차이가 별로 크지 않죠.  

고객 등급	매출	매출 기여도
VIP	2.8억 원	28%  
골드	2.5억 원	25%  
실버	2.5억 원	25%  
브론즈	2.2억 원	22%  
합계	10억 원	100%  

 가전제품의 특성을 고려해 Monetary 가중치를 0.6으로 높이고 Recency, Frequency 가중치는 각각 0.2로 낮춰 다시 세그먼트를 나눴더니 아래처럼 결과가 나왔다고 합니다. VIP 고객의 매출 기여도가 눈에 띄게 높아졌고 아래 등급으로 내려갈수록 매출 기여도가 크게 낮아지고 있죠. 이렇게 되면 VIP 고객이 명백히 핵심 고객인 셈이니 이들을 우선적으로 관리해 주면 될 거예요.    

고객 등급	매출	매출 기여도
VIP	6.4억 원	64%   
골드	2.2억 원	22%  
실버	1.1억 원	11%  
브론즈	0.3억 원	3%

```python
# 나이를 10년 단위로 묶어 '연령대'라는 파생변수 생성 
def age_fun1(x):
    if x >= 80 :
        return '80대 이상'
    elif x >= 70: 
        return '70대'
    elif x >= 60: 
        return '60대'
    elif x >= 50: 
        return '50대'
    elif x >= 40: 
        return '40대'
    elif x >= 30: 
        return '30대'
    elif x >= 20: 
        return '20대'
    else: 
        return '20대 미만'
    
customer_df['age_group'] = customer_df['age'].apply(age_fun1) 
```

```python
# 'age' 컬럼의 위치 찾기
age_col_index = customer_df.columns.get_loc('age')

# 'age' 컬럼 바로 다음 위치에 'age_range' 삽입
# customer_df.pop("age_group"): → 기존 'age_group' 컬럼을 데이터프레임에서 제거하고 값을 반환.
customer_df.insert(loc=age_col_index + 1,
    column='age_group',
    value=customer_df.pop('age_group'),
)

customer_df.head(4)
```

```python
# 연령대별 고객 수 확인
customer_df['age_group'].value_counts()
```

```python
# 수치가 적은 70대와 20대 미만(10대)에 대한 수치를 조정하기 위해 10대와 20를 묶어 20대 이하로,
# 60대, 70대를 60대 이상으로 바꾸어 연령대 값을 새롭게 구하기로 함
# 특정 표본의 수가 적으면 왜곡되기 쉽기 때문
# replace(): 데이터 내의 특정값을 다른 값으로 바꿀 때 사용(딕셔너리 구조로 사용하면 됨)

age_group_replace_dict = {
    '20대 미만': '20대 이하',
    '20대': '20대 이하',
    '60대': '60대 이상',
    '70대': '60대 이상',    
}
customer_df['age_group'] = customer_df['age_group'].replace(age_group_replace_dict)
```

```python
# 이상값 제거(IQR 방식) -> iqr을 구하고, q1 - 1.5*iqr 과 q3 + 1.5*iqr 사이에 있는 데이터만 남기는 방법(and 사용)
income = customer_df['annual_income']
q1 = income.quantile(0.25)
q3 = income.quantile(0.75)
iqr = q3 - q1

condition = ((q1 - 1.5 * iqr <= income) &  (income <= q3 + 1.5 * iqr))
customer_df = customer_df[condition]
```

```python
# 이상값 제거 확인
sns.boxplot(data=customer_df, x='annual_income')
```

### Recency, Frequency, Monetary 등급 매기기 

```python
# 각 항목의 등급 개수 지정: 3개, 등급의 숫자는 클수록 더 긍정적인 것으로 가정. 
# Recency 등급 ↑ -> 더 최근에 구매, Frequency 등급 ↑ -> 더 자주 구매함, Monetary 등급 ↑ -> 돈을 더 많이 씀
# 등급 개수를 미리 선언하여 변수로 지정
num_grades = 3 
```

```python
# customer_df에 grade 등급을 매기기 위해 grade_labels이라는 리스트 변수를 미리 생성
grade_labels = list(range(1, num_grades + 1))
grade_labels
```

```python
# Recency 등급 매기기 
# qcut(): 고객을 비슷한 개수로 그룹(등급) 나누기 위해 qcut()을 사용, 각 그룹에 속하는 고객 수가 최대한 비슷하게 맞춰짐.
# recency 컬럼 값이 작은 고객일수록 최근에 구매를 했다는 뜻이니 Recency 등급을 더 높게 부여해야 함
# [::-1] → 리스트 또는 문자열을 "역순(reverse)"으로 뒤집음 -> 비교: [:-1] → 마지막 요소를 제외한 나머지를 선택
recency_grade = pd.qcut(x=customer_df['recency'], q=num_grades, labels=grade_labels[::-1])
```

```python
# Recency 파생변수 생성
customer_df['recency_grade'] = recency_grade
```

### 가중합을 이용해 RFM 고객 세그먼트 분류하기 

### 고객 지표의 계산:

 Recency, Frequency, Monetary 등급 각각에 가중치를 곱한 뒤 더하면 고객지표가 계산됨. 이 고객지표를 몇 가지 등급으로 나누면 최종 고객 세그먼트를 구할 수 있음. 고객 세그먼트의 등급 개수도 3개로 설정하고 분석을 진행함.  

```python
# 세 항목에 동일한 가중치를 부여
weight = {}
weight['recency'] = 1 / 3
weight['frequency'] = 1 / 3
weight['monetary'] = 1 / 3
```

```python
# 주의: 가중치 적용을 하는 연산을 위해서는 astype() 함수를 이용해 숫자형 데이터로 바꿔 줘야 함.
# customer_df에 Recency, Frequency, Monetary 각 등급이 숫자 1, 2, 3으로 표시되어 있지만,
# 엄밀히 말하면 등급 값은 모두 숫자형 데이터가 아니라 범주형 데이터인 상태
customer_df['rfm_score'] = (
    weight['recency'] * customer_df['recency_grade'].astype('int')
    + weight['frequency'] * customer_df['frequency_grade'].astype('int')
    + weight['monetary'] * customer_df['monetary_grade'].astype('int')
)
```

### 고객 세그먼트 분류 

        고객 지표 구간  등급 
       1 이상 5/3 미만  1등급
       5/3이상 7/3이하  2등급
       7/3초과 3이하    3등급
       

```python
# RFM 세그먼트를 리턴해주는 함수 생성 (apply() 함수 적용)
def rfm_segment_func(x):
    if x < 5 / 3:
        return 1
    elif x <= 7 / 3:
        return 2
    else: 
        return 3
    
customer_df['rfm_segment'] = customer_df['rfm_score'].apply(rfm_segment_func)
```

```python
# RFM 세그먼트 등급표 확인
customer_df['rfm_segment'].value_counts()
```

```python
# rfm_segment 기준으로 매출기여도에 대한 파이차트를 생성
rfm_segment = customer_df.pivot_table(index='rfm_segment', values=['amount_alcohol', 'amount_fruit', 'amount_meat',
                                                'amount_fish', 'amount_snack', 'amount_general', 
                                                'amount_total'], aggfunc='sum').reset_index()
rfm_segment
```

```python
rfm_segment['amount_total'].plot(kind='pie',
                                  autopct='%.1f%%',
                                  labels=[f'{x}등급' for x in grade_labels],
                                  title='RFM 고객 세그먼트별 매출 기여도',
                                  ylabel='',
)
```

```python
# 하지만 앞서 분석에서 Recency 등급은 등급별로 매출 기여도 차이가 거의 없었음 ->
# 따라서 Recency 가중치를 줄이고 다른 항목의 가중치를 높였을 때 고객 세그먼트별 매출 기여도의 차이를 살펴보기로 함.
# 가중치 재설정
weight['recency'] = 0.2
weight['frequency'] = 0.4
weight['monetary'] = 0.4

# 가중합 재계산
customer_df['rfm_score'] = (
    weight['recency'] * customer_df['recency_grade'].astype('int')
    + weight['frequency'] * customer_df['frequency_grade'].astype('int')
    + weight['monetary'] * customer_df['monetary_grade'].astype('int')
)

# RFM 고객 지표 구간화 
customer_df['rfm_segment'] = customer_df['rfm_score'].apply(rfm_segment_func)
```

```python
customer_df['rfm_segment'].value_counts()

# 이전 가중치에 의한 계산보다 1, 2등급의 고객수는 줄고 3등급 세그먼트 고객의 수가 많이 늘어남.
```

```python
# rfm_segment 기준으로 매출기여도에 대한 파이차트를 생성
rfm_segment = customer_df.pivot_table(index='rfm_segment', values=['amount_alcohol', 'amount_fruit', 'amount_meat',
                                                'amount_fish', 'amount_snack', 'amount_general', 
                                                'amount_total'], aggfunc='sum').reset_index()
rfm_segment['amount_total'].plot(kind='pie',
                                  autopct='%.1f%%',
                                  labels=[f'{x}등급' for x in grade_labels],
                                  title='RFM 고객 세그먼트별 매출 기여도',
                                  ylabel='',
)
```

   3등급 세그먼트 고객을 타깃으로 프로모션을 진행할 때 참고할 수 있는 점은 2번 프로모션(주황색 바)의 참여율이 저조했으니 문제점을 분석해봐야하는 과제를 남길 수 있음. 1(파란색 바), 4(빨간색 바), 5(보라색 바)의 프로모션 참여율은 준수하였고, 6번 프로모션(갈색 바)의 참여율은 다른 프로모션에 비해 참여율이 확실히 높기 때문에 6번 프로모션에 대해 집중적으로 분석하는 것도 좋아보임. 이러한 모든 점을 기획해서 프로모션을 기획하는 것을 참고할 수 있을 것 같음.
