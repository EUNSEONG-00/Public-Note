```python
df5 = pd.read_csv('excel/laptops.csv')
df5.head()
```

```python
df5['os'].unique()
```

```python
# catplot: x가 범주형인 데이터를 시각화
sns.catplot(data=df5, x='os', y='price', kind='box')
```

```python
# strip: 모든 데이터의 분포를 볼 수 있음 
sns.catplot(data=df5, x='os', y='price', kind='strip')
```

```python
df5['processor_brand'].unique()
```

```python
# strip 함수에서 어떤 os가 어떤 운영처리장치를 쓰는지 알고 싶어서 세부화
# hue: 색을 의미함 -> 색을 구분해줌
sns.catplot(data=df5, x='os', y='price', kind='strip', hue='processor_brand')
```

```python
# swarm: 점들을 펼쳐줘서 시각화 
sns.catplot(data=df5, x='os', y='price', kind='swarm', hue='processor_brand')
```
