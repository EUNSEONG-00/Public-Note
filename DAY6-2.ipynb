```python
# 모델을 불러와 / 새로운 값을 모델에 입력해 에측
import pickle 
import pandas as pd 
```

```python
# 모델을 불러오기 
model = pickle.load(open('model.sav', 'rb'))
model
```

```python
# 새로운 값 입력하기 (input () )
x1 = input('계약 기간을 입력하시오 : ')
x2 = input('월랜탈비용을 입력하시오 : ')
x3 = input('고객 연령을 입력하시오 : ')
x4 = input('고객 신용 등급을 입력하시오 : ')

input_data = [x1, x2, x3, x4]
```

```python
input_data
```

```python
# 입력받은 개별 값을 데이터 프레임 형태로 변환
new_input = pd.DataFrame(data=[input_data], columns=['Term', 'Amount_Month', 'Age', 'Credit_Rank'])
```

```python
new_input
```

```python
# 입력된 값을 이용해, 예측 수행
model.predict(new_input)
```
