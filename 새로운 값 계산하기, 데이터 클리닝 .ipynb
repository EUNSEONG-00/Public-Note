```python
# 카테고리로 분류 (노트북을 만든 국가별로 분류를 하려고 함)
df3 = pd.read_csv('excel/laptops.csv')
df3.head()
```

```python
brand_nation = {
    'Dell': 'U.S',
    'Apple': 'U.S',
    'Acer': 'Taiwan',
    'HP': 'U.S',
    'Lenovo': 'China',
    'Alienware': 'U.S',
    'Microsoft': 'U.S',
    'Asus': 'Taiwan'
}
```

```python
# .map(): 위에 썻던 딕셔너리 형태에서 왼쪽에 있는 값들을 오른쪽에 있는 값으로 바꿔주는 역할
df3['brand'].map(brand_nation)
```

```python
# 'brand_nation' 이라는 새로운 파생 변수 생성
df3['brand_nation'] = df3['brand'].map(brand_nation)
df3
```

```python
# groupby(): 카테고리별로 비교분석 하는 것을 가능하게 해줌
nation_groups = df3.groupby('brand_nation')
type(nation_groups)
```

```python
nation_groups.count()
```

```python
# 각 컬럼의 최댓값을 비교분석 할 수 있음
nation_groups.max(numeric_only=True)
```

```python
# 각 컬럼의 평균값을 국가별로 비교분석 할 수 있음 
nation_groups.mean(numeric_only=True)
```

```python
# 각 국가별로 첫 번째 값을 보여줌
nation_groups.first()
```

```python
# 각 국가별로 가장 마지막 값을 보여줌
nation_groups.last()
```

```python
nation_groups.plot(kind='box', y='price')
```

```python
nation_groups.plot(kind='hist', y='price')
```

```python
# 이상점(outlier) 확인 과정
df6 = pd.read_csv('excel/beer.csv', index_col=0)
df6.head()
```

```python
# 'adv': 도수라는 변수를 활용한 이상점
df6.plot(kind='box', y='abv')
```

```python
df6['abv'].describe()
```

```python
df6['abv'].quantile(0.25)
```

```python
df6['abv'].quantile(0.75)
```

```python
# 사분위수 범위 설정 후 이상점 찾는 과정
q1 = df6['abv'].quantile(0.25)
q3 = df6['abv'].quantile(0.75)
iqr = q3 -q1
```

```python
condition = (df6['abv'] < q1 - 1.5 * iqr) | (df6['abv'] > q3 + 1.5 * iqr)
```

```python
df6[condition]
# 데이터 분석 결과:
# 963번은 소주로 확인, 1856번은 보드카, 2250은 맥주는 맞으나 수치를 잘못입력한 것으로 예측됨
```

```python
df6.loc[2250, 'abv'] = 0.055
```

```python
df6.loc[2250]
```

```python
condition = (df6['abv'] < q1 - 1.5 * iqr) | (df6['abv'] > q3 + 1.5 * iqr)
df6[condition]
```

```python
df6[condition].index
```

```python
# 나머지 두 인덱스는 지우는 것이 합당함
df6.drop(df6[condition].index, inplace=True)
```

```python
df6[condition]
```

```python
# 이상값 없는 깔끔한 박스플롯 도출 가능
df6.plot(kind='box', y='abv')
```
