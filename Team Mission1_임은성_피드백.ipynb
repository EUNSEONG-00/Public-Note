```python
# 1번  / 윤년 : 4로 나누어 떨어지고, 대신 100으로 나누어 떨어지면 안된다 or 400으로 나누어 떨어진다 
def leapYear(x):
    cond = (x % 4 == 0) & (x % 100 != 0) | (x % 400 ==0)
    dic = {True : '윤년', False : '평년'}
    return dic[cond]
leapYear(2023)
```

```python
# 9번 피드백

# 위의 코드는 cond1을 지정해주는 이유를 이해하지 못한 것으로 보입니다.

# cond1은 'Channel'이 '대형마트A'에 해당하는 열만 
# 필터링하기 위해 만든 조건입니다.

# cond1 이라는 필터를 만든 후 .loc 함수로 필터링을 하고
# sort_values와 이후의 코드를 사용해야합니다.



# 9. 유입경로(Channel)에서 '대형마트A'로 유입된 고객들 중, 
# 연체건수(Overdue_count)가 높은 상위 30명의 계약 명단을 추출하여 
# result4.xlsx 저장하시오.
```

```python
# 9. 유입경로(Channel)에서 '대형마트A'로 유입된 고객들 중, 

# c1은 condition1 을 뜻합니다
c1 = df1['Channel'] == '대형마트A'
c1
# c1은 아래와 같이 True / False 를 가진 열입니다
# 대형마트A인 경우 True를, 그렇지 않은 경우는 False입니다
```

```python
c2 = df1.loc[c1]
c2
# .loc 함수에 [c1]을 집어넣으면 True가 있는 열만 가져옵니다
# 이 데이터프레임을 c2로 저장합니다
```

```python
c2['Channel'].value_counts()

# c2의 Channel에는 대형마트만 있는 것을 확인할 수 있습니다
```

```python
# 연체건수(Overdue_count)가 높은 상위 30명의 계약 명단을 추출하여 

c3 = c2.sort_values(by = 'Overdue_count', ascending = False).head(30)

# .sort_values 함수를 사용해 오름차순이나 내림차순으로 출력합니다

# by = 에는 어느 컬럼을 내림/오름차순으로 표현할지를 정해줍니다

# ascending = False는 내림차순으로 표현하는 옵션입니다

# .head(30)는 상위 30명만까지만 출력합니다
```

```python
# result4.xlsx 저장하시오.

c3.to_excel('result4.xlsx')
```
