```python
# 제품군('Product_Type') 별로 월랜탈비용('Amount_Month')의 평균을 확인
df1.pivot_table(index='Product_Type', values='Amount_Month', aggfunc='mean')
```

```python
# 제품군('Product_Type') 별로 월랜탈비용('Amount_Month')의 평균과 합을 확인
df1.pivot_table(index='Product_Type', values='Amount_Month', aggfunc=['mean', 'sum'])
```

```python
# 제품군('Product_Type') 별로 월랜탈비용('Amount_Month')과 계약기간(Term)의 평균을 확인
df1.pivot_table(index='Product_Type', values=['Amount_Month','Term'], aggfunc='mean')
```

```python
# 제품군('Product_Type')과 계약유형(Sales_Type) 별로 월랜탈비용('Amount_Month')과 계약기간(Term)의 평균을 확인
df1.pivot_table(index=['Product_Type', 'Sales_Type'], 
                values=['Amount_Month','Term'], aggfunc='mean')
```

```python
# 제품군('Product_Type') 별로 월랜탈비용('Amount_Month')의 합을 확인, 열을 계약유형(Sales_Type)별로 정리
df1.pivot_table(index='Product_Type', values='Amount_Month', 
                aggfunc='sum', columns='Sales_Type')
```

```python
df1.pivot_table(index='Product_Type', values='Amount_Month', 
                aggfunc='sum', columns='Sales_Type', fill_value=0)
# fill_value : pivot.table 계산을 통해 발생한 결측값에 원하는 값을 입력 
```

```python
# 날짜 데이터 변환 
df1['Datetime_dt'] = pd.to_datetime(df1['Datetime'])  # -> 이렇게 한다고 해서 df1에 Datetime이 바로 덮어서 전환되는 것은 아니다. 
# Datetime 데이터를 날짜형식의 데이터로 변환하여 df1 데이터 프레임에 새로운 Datetime_dt 변수로 선언함
```

```python
df1['Year'] = df1['Datetime_dt'].dt.year    # 날짜형식의 데이터에서 연도 추출 
df1['Month'] = df1['Datetime_dt'].dt.month  # 월 추출
df1['Day'] = df1['Datetime_dt'].dt.day      # 일 추출 
df1['Day_of_week'] = df1['Datetime_dt'].dt.day_name()   # 요일 추출
df1['Week'] = df1['Datetime_dt'].dt.isocalendar().week  # 주차 추출 
```

```python
df1
```

```python
# 연도 별, 제품군 별 월랜탈비용의 합을 계산하여 데이터파일로 저장
df1.pivot_table(index='Product_Type', values='Amount_Month', 
               columns= 'Year', aggfunc='sum')
```

```python
# 파생 변수 생성
# 월랜탈비용(Amount_Month)와 계약기간(Term)을 곱하여, 총계약금액(Total_Amount)을 계산
df1['Total_Amount'] = df1['Amount_Month'] * df1['Term'] 
```

```python
# 비교구문과 loc함수를 활용해 데이터를 그룹화 ex) 40대 이상은 40대, 80대 이상은 80대로 묶기 
cond1 = (df1['Age']>=70)
cond2 = (df1['Age']>=60)
cond3 = (df1['Age']>=50)
cond4 = (df1['Age']>=40)
cond5 = (df1['Age'] < 40)

df1.loc[cond5, '연령대'] = '30대이하'
df1.loc[cond4, '연령대'] = '40대'
df1.loc[cond3, '연령대'] = '50대'
df1.loc[cond2, '연령대'] = '60대'
df1.loc[cond1, '연령대'] = '70대이상'
```

```python
# apply( ) : 파이썬 함수의 개념을 활용해 데이터를 처리
def func1(x):
    if x >= 70 : 
        return '70대이상'
    elif x >= 60 : 
        return '60대'
    elif x >=  50 : 
        return '50대'
    else: 
        return '40대이하'


df1['Age'].apply(func1)
```

```python
func1(400)
```

```python
def func2(x):
    return str(x)[0] + '0대'        # str() : 숫자 데이터를 문자데이터로 전환시켜준다


df1['연령대'] = df1['Age'].apply(func2)
```

 - 결측값 확인 및 처리 (Missing Value => NaN (Not A Number))
 
 - 결측값 제거 : 결측값이 존재하는 행을 제거 (보통 가장 많이 쓴다 / 결측치를 대치해서 어떤 수를 집어넣는다면 전체 데이터의
                 왜곡이 심해지기 때문이다)
 - 결측값 대치 : 결측값을 다른 값으로 채워넣는 방법 (결측치의 개수가 너무 많다면, 대치를 함)
    - 단순대치 
    - 알고리즘 대치 : 머신러닝 기법 중 하나, 주변 값과 유사한 값을 이용해 대치를 함 
    - 보간법 : ex) 공장에서 품질 데이터 - 4시간, 공정 데이터 - 10분 단위 추출 // -> 그 사이에 있는 데이터의 개수만큼 데이터의 간격을 
               나눠서 대치하는 방법

 # 결측치를 나타내는 값

```python
# 결측값 대치 (단순대치) / fillna() : 특정 값으로 결측값을 대치
df1['Credit_Rank_clean1'] = df1['Credit_Rank'].fillna(0)
df1['Credit_Rank_clean1']
```

```python
# 결측값에 평균 대치 (단순대치)
df1['Credit_Rank'].fillna( df1['Credit_Rank'].mean() )
```

```python
# 결측값에 중앙값 대치 (단순대치)
df1['Credit_Rank'].fillna( df1['Credit_Rank'].median() )
```
