```python
# jointplot: 변수와 변수간의 연관성과 각 변수의 히스토그램 분포를 파악할 수 있다. 
sns.jointplot(data=basic_info, x='Height', y='Weight')
```

```python
# 숫자가 아닌 열을 제외하고 모든 상관계수 계산
df.select_dtypes(include=["number"]).corr()
```

```python
# 나이 변수와 관련한 상관계수들을 내림차순으로 정렬 
df.select_dtypes(include=["number"]).corr()['Age'].sort_values(ascending=False)
```
